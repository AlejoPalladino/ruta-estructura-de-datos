### **Ejercicio 1**

En un trabajo estamos implementando el software de administración de ventas de la línea de cajas de un supermercado. Para esto tenemos que crear el TDA **CajaDeSupermercado** que contiene:

- Una estructura de datos que contiene el historial de ventas que se cobraron en esa caja, donde una venta consiste en un monto y el horario en el que fue realizada. En esta estructura se accede primero solo a la primera venta que se hizo en el día. La estructura debe ser dinámica. **NO** se puede usar una Lista.
- El nombre de la persona que administra la caja
- El número de la caja en la línea

Implementar las siguientes operaciones en el TDA CajaDeSupermercado:

- **\_\_init\_\_(self, nombre:str, numero:int)**: que crea una variable de tipo CajaDeSupermercado
- **agregarVenta(self, horario:str, montoDeVenta:float) → None**: que guarda la venta realizada. Por simplicidad asumir que el horario es un string
- **cantidadDeVentasGrandes(self, montoMinimo:float) → int**: que calcula la cantidad de ventas que se realizaron en la caja que sean mayores al monto que se recibe como parámetro. El historial de ventas NO debe ser modificado.
- **ultimasNVentas(self, N:int) → np.array[ventas]**: que retorna un arreglo conteniendo los montos de las últimas N ventas que se cobraron en la caja. El historial de ventas NO debe ser modificado.
- **vaciarHistorial(self) → None**: que vacía el historial de ventas retornando la cantidad de ventas eliminadas.
- **montoVentasHorario(self, horario:str) → float**:  que retorna el importe de la venta realizada en ese horario.

### **Ejercicio 2**

En una biblioteca se proponen controlar los préstamos de libros que hacen diariamente. Para esto nos piden hacer un software que administre la mesa de entradas. Crear el TDA **MesaDeEntradas** que contiene:

- Una estructura de datos que contiene a los Libros que se prestaron hasta ese momento, donde un libro consiste en un código (número entero), un título y un género (3 géneros posibles: Teatro, Poesía o Narración). En esta estructura se accede primero al último libro que se prestó. La estructura debe ser dinámica. NO se puede usar una Lista.
- El nombre de la persona que administra la mesa de entradas.
- El número de piso de la mesa entrada en el edificio de la biblioteca.

Implementar las siguientes operaciones del TDA **MesaDeEntradas**:

- **\_\_init\_\_(self, nombre:str, numero:int)**: que crea una variable de tipo MesaDeEntradas
- **prestarLibro(self, codigoLibro:int, tituloLibro:str, generoLibro:str) → None**: que recibe los datos de un libro y lo almacena en la estructura de libros prestados.
- **verificarPrestamo(self, codigoLibro:int) → bool**: que retorna True si el libro con el código recibido fue prestado y False en caso contrario. NO se debe modificar la estructura que almacena los libros prestados.
- **SacarLibrosDeGenero(self, generoLibro:str) → int**: que elimina de la estructura los libros del género que se recibe por parámetro y retorna la cantidad.
- **primerosLibros(self, N:int) → np.array[str]**: que retorna un arreglo conteniendo los títulos de los primeros N libros que se prestaron. NO se debe modificar la estructura que almacena los libros prestados.
- **vaciarMesa(self) → None**: que vacía la estructura de libros prestados retornando la cantidad de libros eliminados.

Pero atención, cuando llega un paquete nuevo, este debe ser intercalado de manera de respetar el orden del peso.

Modelar el TDA **Correo** con las siguientes operaciones:

- **juntarCorreos(self, otroCorreo) → None**: que recibe otra instancia del TDA, y **apila en self** los **paquetes de self** combinados con los de **otroCorreo**, de manera que no se aplasten. otroCorreo debe quedar intacto luego de la operación. Se recomienda usar pilaPaquetes(). Aprovechar que otroCorreo ya viene ordenado.


### **Ejercicio 4**

Estamos implementado un software para administrar una red de impresoras y para eso tenemos que guardar el conjunto de trabajos que los usuarios envían para imprimir.
Implementar el TDA **RedImpresoras** que modela una red de impresoras, almacenando los trabajos pendientes y tambien los trabajos que están actualmente en proceso de impresión. Al almacenar los trabajos pendientes nos interesa conservar el orden, es decir el último trabajo que ingresa debe ser el último en imprimirse (para almacenar los trabajos en ejecución no hace falta conservar el orden).

Se recomienda modelar primero el TDA Trabajo, que tendrá un nombre de archivo, un tamaño de papel (“Oficio”, “A4” o “Carta”) y un tiempo estimado de impresión en segundos.
Para este ejercicio es obligatorio utilizar los TDAs Pila y/o Cola para almacenar los trabajos en la red. Pueden usar todas las operaciones vistas en clase.

Implementar el TDA RedImpresoras que contenga al menos las siguientes operaciones:

- **\_\_init\_\_(self)**: que crea una variable de tipo RedImpresoras.
- **agregarTrabajo(self, nombreArchivo:str, tamaño:str, tiempo:int) → None**: que crea y agrega un trabajo a los pendientes.
- **mandarAmprimir(self, tiempoMax:int) → str**: que busca el primer trabajo pendiente con un tiempo de impresión menor a tiempoMax y devuelve el nombre del archivo a imprimir. La función debe pasar el trabajo de estar entre los “pendientes” a los de “en proceso de impresión”.
- **trabajoTerminado(self, nombreArchivo:str) → None**: que elimina el trabajo del nombreArchivo de los trabajos en proceso, si hay dos trabajos con el mismo nombre de archivo elimina el primero que encuentra.
- **trabajosTamaño(self, N:int, tamaño:str) → np.array[str]**: que devuelve un arreglo con los nombres de los primeros N archivos que están pendientes de impresión con el tamaño de papel que se recibe como parámetro.
